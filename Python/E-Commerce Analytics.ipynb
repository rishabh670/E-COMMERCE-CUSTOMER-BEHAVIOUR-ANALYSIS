```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df.describe()
```

```python
df = df[df["Quantity"] > 0]
df = df[df["UnitPrice"] > 0]
```

```python
df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]
```

```python
df.to_csv("../Dataset/cleaned_data.csv", index=False)
```
